## 🧠 Aliased Tables ile JOIN (Table Aliasing)

Bu örnekte, `countries` ve `economies` tablolarını **alias (takma ad)** kullanarak birleştiriyoruz.

---

### ❗ Neden Alias Kullanılır?

Farklı tablolarda aynı isimli kolonlar olabilir. Örneğin:

* `countries.name`
* `economies.name`

Bu durumda SQL şu hatayı verir:

```text id="b2r4qk"
column reference "name" is ambiguous
```

👉 Çözüm: Tablo alias kullanmak

---

## 📌 SQL Sorgusu

```sql id="y3k9pd"
%%sql

-- countries tablosuna "c", economies tablosuna "e" alias veriyoruz
SELECT 
    c.code AS country_code,   -- ülke kodu
    c.name,                  -- ülke adı
    e.year,                  -- yıl bilgisi
    e.inflation_rate         -- enflasyon oranı

-- Ana tablo: countries (alias: c)
FROM countries AS c

-- economies tablosu ile INNER JOIN (alias: e)
INNER JOIN economies AS e

-- İki tabloyu country code üzerinden eşleştir
ON c.code = e.code;
```

---

## 🧠 Açıklamalar

* `AS c`, `AS e`
  → Tablolara kısa isim verir (alias)

* `c.code`, `e.year`
  → Hangi kolonun hangi tablodan geldiğini açıkça belirtir

* `INNER JOIN`
  → Sadece eşleşen kayıtları getirir

---

## 🎯 Bu Sorgu Ne Yapar?

* Ülkeleri ve ekonomik verilerini birleştirir
* Her ülke için:

  * yıl
  * enflasyon oranı
    bilgilerini getirir

---

## 🚀 Neden Önemli?

Bu yapı:

* Gerçek projelerde **zorunlu**
* Özellikle çok tablo ile çalışırken **hayat kurtarır**
* Okunabilir ve profesyonel SQL yazmanı sağlar

---


````python
%%sql

-- countries tablosuna "c", economies tablosuna "e" alias (takma ad) veriyoruz
-- Bu sayede uzun tablo isimlerini tekrar tekrar yazmak zorunda kalmayız

-- Seçilecek kolonlar:
SELECT 
    c.code AS country_code,   -- countries tablosundaki code kolonunu al ve adını country_code yap
    c.name,                  -- ülke adı (countries tablosundan)
    e.year,                  -- yıl bilgisi (economies tablosundan)
    e.inflation_rate         -- enflasyon oranı (economies tablosundan)

-- Ana tablo: countries
FROM countries AS c

-- economies tablosu ile INNER JOIN yapıyoruz
-- INNER JOIN → sadece eşleşen kayıtları getirir
INNER JOIN economies AS e

-- İki tabloyu country code üzerinden eşleştiriyoruz
-- countries.code = economies.code
ON c.code = e.code
LIMIT 100;
````

## 🧠 USING ile JOIN (countries & languages)

Bu örnekte, `countries` ve `languages` tablolarını **USING** anahtar kelimesi ile birleştiriyoruz.

---

### ❗ USING Nedir?

Eğer iki tabloda join yapılacak kolonun adı **aynıysa**:

```sql
ON countries.code = languages.code
```

yerine daha kısa olarak:

```sql
USING(code)
```

yazabiliriz.

---

## 📌 SQL Sorgusu

```sql id="p7k3sd"
%%sql

-- countries (c) ve languages (l) tablolarını birleştiriyoruz
SELECT 
    c.name,        -- ülke adı
    l.language,    -- konuşulan dil
    l.official     -- dil resmi mi? (True/False)

FROM countries AS c

-- USING ile INNER JOIN
-- code kolonu her iki tabloda da aynı isimde olduğu için USING kullanıyoruz
INNER JOIN languages AS l
USING(code);
```

---

## 🧠 Açıklamalar

* `USING(code)`
  → Her iki tabloda da bulunan `code` kolonuna göre join yapar

* `c.name`, `l.language`
  → Hangi kolonun hangi tablodan geldiği açıkça belirtilir

* `l.official`
  → Dilin resmi olup olmadığını gösterir

---

## 🎯 Bu Sorgu Ne Yapar?

* Ülkeler ile konuşulan dilleri birleştirir
* Her ülke için:

  * konuşulan dil
  * dilin resmi olup olmadığı
    bilgilerini getirir

---

## 🚀 Neden Önemli?

* Daha kısa ve temiz SQL yazımı sağlar
* Aynı isimli kolonlar için idealdir
* Gerçek projelerde sık kullanılır

---


````python
%%sql

-- countries tablosuna "c", languages tablosuna "l" alias veriyoruz
-- Amaç: daha kısa ve okunabilir sorgu yazmak

SELECT 
    c.name AS country,   -- ülke adı (countries tablosundan)
    l.name AS language,  -- dil adı (languages tablosundan)
    official             -- dil resmi mi? (True/False)

FROM countries AS c

-- languages tablosu ile INNER JOIN
-- INNER JOIN → sadece eşleşen kayıtları getirir
INNER JOIN languages AS l

-- USING kullanımı:
-- Eğer iki tabloda da kolon adı aynıysa (burada: code),
-- ON c.code = l.code yazmak yerine USING(code) kullanabiliriz
USING(code)
LIMIT 100;
````

## 🧠 Multiple JOIN (countries, populations, economies)

Bu sorguda, üç farklı tabloyu birleştirerek:

* ülke adı
* doğurganlık oranı
* yıl
* işsizlik oranı

bilgilerini tek bir sonuçta elde ediyoruz.

---

### 📌 SQL Sorgusu

```sql id="r8k3pd"
%%sql

-- countries (c), populations (p) ve economies (e) tablolarını birleştir

SELECT 
    c.name,                -- ülke adı
    p.fertility_rate,      -- doğurganlık oranı
    e.year,                -- yıl bilgisi
    e.unemployment_rate    -- işsizlik oranı

-- Ana tablo: countries
FROM countries AS c

-- populations tablosu ile INNER JOIN
-- country_code üzerinden eşleşme
INNER JOIN populations AS p
ON c.code = p.country_code

-- economies tablosu ile ikinci JOIN (chain join)
INNER JOIN economies AS e
ON c.code = e.code;
```

---

### 🧠 Açıklamalar

* `INNER JOIN`
  → Sadece eşleşen kayıtları getirir

* `c.code = p.country_code`
  → countries ↔ populations ilişkisi

* `c.code = e.code`
  → countries ↔ economies ilişkisi

* **Chain JOIN (çoklu join)**
  → Birden fazla tabloyu tek sorguda birleştirir

---

### 🎯 Bu Sorgu Ne Yapar?

Her ülke için:

* doğurganlık oranı
* belirli yıllardaki işsizlik oranı

bilgilerini birlikte analiz etmeni sağlar

---

### 🚀 Neden Önemli?

Bu yapı:

* Gerçek veri analizi projelerinin temelidir
* Çoklu veri kaynaklarını birleştirmeni sağlar
* SQL’de ileri seviye sorguların başlangıcıdır

---


````python
%%sql

-- Ülke adı ve doğurganlık oranını seç
SELECT 
    c.name,           -- countries tablosundan ülke adı
    p.fertility_rate  -- populations tablosundan doğurganlık oranı

-- Ana tablo: countries
FROM countries AS c

-- populations tablosu ile INNER JOIN
-- country_code üzerinden eşleştirme yapılır
INNER JOIN populations AS p
ON c.code = p.country_code
LIMIT 100;
````

````python
%%sql

-- Ülke adı, yıl, doğurganlık oranı ve işsizlik oranını seç
SELECT 
    c.name,                 -- countries tablosundan ülke adı
    e.year,                 -- economies tablosundan yıl bilgisi
    p.fertility_rate,       -- populations tablosundan doğurganlık oranı
    e.unemployment_rate     -- economies tablosundan işsizlik oranı

-- Ana tablo: countries
FROM countries AS c

-- populations tablosu ile INNER JOIN
-- countries.code = populations.country_code
INNER JOIN populations AS p
ON c.code = p.country_code

-- economies tablosu ile ikinci JOIN
INNER JOIN economies AS e

-- USING(code):
-- economies.code ile countries.code üzerinden eşleşme
-- (ON c.code = e.code ile aynı)
USING (code)
LIMIT 100;
````

## 🧠 Multi-Table JOIN Hatasını Düzeltme (Year Eşleşmesi)

Önceki sorguda, `economies` tablosu sadece `code` üzerinden join edildiği için
farklı yıllara ait veriler birbirine karıştı.

👉 Sonuç: Aynı ülke için gereksiz çoğaltılmış satırlar oluştu.

---

## ❗ Problemin Sebebi

```sql id="err1"
ON c.code = e.code
```

Bu join:

* Ülkeyi eşleştirir ✔️
* Ama yılı dikkate almaz ❌

👉 Bu yüzden:

* 2010 doğurganlık → 2015 işsizlik ile eşleşebilir

---

## ✅ Doğru Çözüm

Hem **ülke kodu** hem de **yıl** eşleşmeli:

```sql id="fix1"
ON c.code = e.code 
AND p.year = e.year
```

---

## 📌 Düzeltilmiş SQL Sorgusu

```sql id="final1"
%%sql

SELECT 
    c.name,                 
    e.year,                 
    p.fertility_rate,       
    e.unemployment_rate     

FROM countries AS c

INNER JOIN populations AS p
ON c.code = p.country_code

INNER JOIN economies AS e
ON c.code = e.code
AND p.year = e.year;
```

---

## 🎯 Bu Sorgu Ne Yapar?

* Aynı ülke ve aynı yıl için:

  * doğurganlık oranı
  * işsizlik oranı
    verilerini doğru şekilde eşleştirir

---

## 🧠 Öğrenmen Gereken Kritik Nokta

👉 **Time-series verilerde JOIN yaparken sadece ID yetmez**

Mutlaka:

* `code` (ülke)
* `year` (zaman)

gibi tüm anahtarlar birlikte kullanılmalıdır.

---

## 🚀 Sonuç

* Tekrarlanan satırlar giderildi ✔️
* Veri doğru eşleşti ✔️
* Analiz güvenilir hale geldi ✔️

---


````python
%%sql

-- Ülke adı, yıl, doğurganlık oranı ve işsizlik oranını getir
SELECT 
    c.name,                 -- countries tablosundan ülke adı
    e.year,                 -- economies tablosundan yıl
    p.fertility_rate,       -- populations tablosundan doğurganlık oranı
    e.unemployment_rate     -- economies tablosundan işsizlik oranı

-- Ana tablo: countries
FROM countries AS c

-- populations ile join (ülke bazlı)
INNER JOIN populations AS p
ON c.code = p.country_code

-- economies ile join (ülke + yıl bazlı)
INNER JOIN economies AS e
ON c.code = e.code
AND p.year = e.year
LIMIT 100;   -- yıl da eşleşmeli (doğru sonuç için)
````
