# Veri Modellemesi Nedir?

Veri modelleme (Data Modeling), gerçek dünyadaki iş süreçlerini, varlıkları (entities), ilişkileri ve kuralları sistematik bir veri yapısına dönüştürme sürecidir.

Başka bir ifadeyle:

> “Gerçek dünyayı veritabanının anlayabileceği şekilde temsil etme sanatı.”

---

# Veri Modellemesinin Büyük Resimdeki Yeri

```text
İş Süreçleri
      ↓
Veri Modelleme
      ↓
Veritabanı Tasarımı
      ↓
ETL / ELT Süreçleri
      ↓
Data Warehouse
      ↓
Analytics & BI
      ↓
Machine Learning
```

---

# Veri Modelleme Neden Kritik Öneme Sahiptir?

İyi tasarlanmış bir veri modeli:

- Veri tekrarını azaltır
- Veri tutarsızlığını engeller
- Analitik süreçleri kolaylaştırır
- Performansı iyileştirir
- Ölçeklenebilirlik sağlar
- Veri kalitesini artırır
- İş kurallarını standartlaştırır
- Bakımı kolaylaştırır

Kötü tasarlanmış bir veri modeli ise:

- Karmaşık sorgular
- Yavaş çalışan sistemler
- Veri tutarsızlığı
- Sürekli teknik borç
- Bakım maliyeti
- Analitik problemlerde güvensizlik

gibi ciddi sorunlara yol açabilir.

# Veri Modellemede Temel Yaklaşımlar

Bu bölümde üç ana yaklaşım incelenecek:

## 1. Relational Modeling (İlişkisel Modelleme)

- Normalize yapı
- Transaction odaklı sistemler
- Veri tutarlılığı önceliği
- OLTP sistemlerinde yaygın kullanım

---

## 2. Dimensional Modeling (Boyutsal Modelleme)

- Analitik sorgular için optimize edilir
- Star schema / Snowflake schema
- BI ve veri ambarlarında yaygın kullanım
- Hızlı raporlama avantajı

---

## 3. Data Vault

- Modern veri ambarı yaklaşımı
- Esnek ve ölçeklenebilir yapı
- Tarihsel veri takibi güçlüdür
- Büyük kurumsal sistemlerde tercih edilir

---

# Veri Modellemede Temel Trade-Off Kavramı

Veri modelleme çoğu zaman bir denge kurma işidir.

Örneğin:

| Amaç | Avantaj | Dezavantaj |
|---|---|---|
| Yüksek normalizasyon | Veri tekrarını azaltır | Sorgular karmaşıklaşabilir |
| Denormalizasyon | Performansı artırabilir | Veri tekrarı oluşabilir |
| Esneklik | Yeni ihtiyaçlara uyum sağlar | Sistem karmaşıklaşabilir |
| Basitlik | Yönetim kolaylığı sağlar | Bazı ihtiyaçları karşılamayabilir |

---

# Trade-Off Mantığının Görselleştirilmesi

```text
Yüksek Normalizasyon
        │
        ├──► Daha Az Veri Tekrarı
        │
        └──► Daha Karmaşık JOIN Yapıları


Denormalizasyon
        │
        ├──► Daha Hızlı Analitik Sorgular
        │
        └──► Daha Fazla Veri Tekrarı
```

Veri bilimi projelerinde model performansı çoğu zaman:

- algoritmadan değil,
- veri kalitesinden,
- veri organizasyonundan,
- veri erişilebilirliğinden

etkilenir.

- veri sızıntısına (data leakage) neden olabilir.

Bu nedenle veri modelleme yalnızca veri mühendislerinin değil:

- veri analistlerinin,
- BI geliştiricilerin,
- machine learning engineer’ların,
- data scientist’lerin

de anlaması gereken temel bir konudur.

Downstream sistem:

# Veri Güncellemesindeki Kritik Sorular

Yeni kolon eklendiğinde şu sorular ortaya çıkar:

## 1. Eski kayıtlar ne olacak?

Örneğin:

| customer_id | marital_status |
|---|---|
| 101 | NULL |
| 102 | NULL |

Burada seçenekler:

- NULL bırakmak,
- başka sistemden veriyi almak,
- geçmiş verileri doldurmak (backfill)

olabilir.

---

# Backfill Nedir?

Backfill:

> Geçmiş verileri sonradan doldurma işlemidir.

Örneğin:

- eski müşteri kayıtlarını başka sistemlerden okuyup,
- marital_status alanını doldurmak

bir backfill işlemidir.

---

# Veri Modelinde Esneklik Neden Kritik?

Esnek olmayan modellerde:

- küçük değişiklikler büyük maliyet yaratır,
- sistem bağımlılıkları kırılır,
- veri kalitesi bozulabilir,
- bakım maliyeti artar.

Bu nedenle iyi veri modelleri:

```text
Değişiklikleri minimum maliyetle absorbe edebilmelidir.
```

✓ eksik veri bırakmamalı

✓ genişlemeye uygun olmalı  
✓ downstream sistemleri koruyabilmeli  

---

# Enforcement of Business Rules  
# İş Kurallarının Uygulanması

Bir örnekle başlayalım.

Yeni bir ev inşa ettiğinizi düşünün.

Bu süreçte:

- yerel yasalar,
- güvenlik kuralları,
- belediye yönetmelikleri,
- mühendislik standartları

gibi kurallara uymak zorundasınızdır.

Bu kurallar:

- projenin kapsamını belirler,
- maliyeti etkiler,
- güvenliği sağlar,
- hataları önler.

Kuralları en başta bilmek:

```text
Daha sağlıklı planlama yapılmasını sağlar.
```

---

# Veri Sistemlerinde İş Kuralları

Aynı mantık veri sistemleri için de geçerlidir.

Bir veri modeli:

> İçerisinde tuttuğu verilerle ilgili iş kurallarını desteklemeli ve uygulamalıdır.

Çünkü veri modeli yalnızca veri saklayan teknik bir yapı değildir.

Aynı zamanda:

```text
İş dünyasının kurallarını temsil eden yapıdır.
```

---

# İş Kuralları Neler Olabilir?

Örnek iş kuralları:

| İş Kuralı | Açıklama |
|---|---|
| Her müşteri benzersiz olmalı | Duplicate kayıt engellenir |
| Her faturanın tek ID’si olmalı | Veri tutarlılığı sağlanır |
| Sipariş tarihi boş olamaz | Veri kalitesi korunur |
| Çalışan maaşı negatif olamaz | İş mantığı korunur |

---

# İş Kuralları Neden Önemlidir?

Bazı iş kuralları:

- operasyonları kolaylaştırmak,
- veri kalitesini artırmak,
- standartlaşma sağlamak

için oluşturulur.

Bazıları ise:

- yasal zorunluluklar,
- regülasyonlar,
- sektör standartları

nedeniyle gereklidir.

---

# Veri Modeli İş Kurallarını Nasıl Uygular?

Veri modeli:

- primary key,
- foreign key,
- unique constraint,
- not null constraint,
- check constraint

gibi yapılarla iş kurallarını enforce edebilir.

---

# Örnek: Fatura Numarası

Bir fabrikada her faturanın benzersiz bir numarası olması gerektiğini düşünelim.

Bu durumda veri modelinde:

```sql
invoice_id UNIQUE
```

tanımlamak iş kuralını sistem seviyesinde enforce eder.

Bu yaklaşım:

- duplicate veri oluşmasını engeller,
- operasyonel hataları azaltır,
- veri güvenilirliğini artırır.

---

# Yanlış Modelleme Problemi

Eğer veri modeli iş kurallarını doğru temsil etmezse:

- veri kalitesi bozulabilir,
- tutarsız kayıtlar oluşabilir,
- operasyonel hatalar ortaya çıkabilir,
- sistem güvenilirliği azalabilir.

Daha kötüsü:

```text
Hatalı veri sistemi zamanla organizasyonun “gerçeği” haline gelir.
```

---

# İş Kuralları Zamanla Değişebilir

Önemli bir gerçek:

```text
Business rules are not static.
```

Yani iş kuralları zaman içerisinde değişebilir.

Örneğin:

- yeni regülasyonlar çıkabilir,
- şirket politikaları değişebilir,
- yeni ürünler eklenebilir.

Bu nedenle veri modeli:

- yalnızca bugünkü kuralları değil,
- değişebilecek kuralları da

olabildiğince desteklemelidir.

---

# Business Stakeholder Katılımı

İyi veri modeli yalnızca teknik ekip tarafından tasarlanmaz.

İş birimlerinin (business stakeholders) sürece dahil edilmesi gerekir.

Çünkü iş kurallarını en iyi bilen kişiler çoğu zaman:

- finans ekipleri,
- ürün yöneticileri,
- hukuk ekipleri,

# İş Birimlerini Sürece Dahil Etmenin Faydaları

| Belirsizlik azalır | Kurallar netleşir |
| Gelecek değişiklikler öngörülebilir | Sistem daha dayanıklı olur |
| Veri kalitesi artar | Hatalı kayıtlar azalır |

---

# Veri Modellemede Temel Gerçek

İyi bir veri modeli:

```text
Sadece veri saklamaz.
İş dünyasının mantığını da taşır.
```

# Minimizing Redundancy

```text
“Tüm gerekli veriye sahip olmak”
```

değil, aynı zamanda bunu:

```text
“Minimum gereksiz tekrar ile yapmak”
```

olmalıdır.

---

# Gerçek Hayattan Bir Analoji

- gereksiz tekrarlar maliyet oluşturur.

```text
İki banyo olup salon olmaması mantıklı değildir.
```

Çünkü:

- bazı alanlar gereksiz yere tekrar eder,
- bazı kritik ihtiyaçlar eksik kalır,
- kullanım deneyimi bozulur.

# Stability and Flexibility  
# Kararlılık ve Esneklik

Bir evin bahçesinde sürekli parti düzenlediğinizi düşünelim.

Eğer bu yalnızca tek seferlik bir etkinlikse:

```text
Geçici bir çadır yeterli olabilir.
```

Ancak bu tür etkinlikleri sürekli yapıyorsanız:

```text
Daha kalıcı ve dayanıklı bir yapı inşa etmek gerekir.
```

Veri modelleme dünyasında da benzer bir durum vardır.

---

# İş Dünyası Sürekli Değişir

Gerçek dünyada iş gereksinimleri statik değildir.

Zamanla:

- yeni ürünler ortaya çıkar,
- yeni bölgelerde operasyon başlar,
- yeni regülasyonlar gelir,
- yeni müşteri tipleri oluşur,
- yeni veri kaynakları eklenir.

Dolayısıyla veri modeli:

```text
Değişime dayanıklı olmalıdır.
```

---

# Stability (Kararlılık) Nedir?

Stability:

> Veri modelinin zaman içerisindeki değişikliklere rağmen çalışmaya devam edebilme yeteneğidir.

İyi bir veri modeli:

- küçük değişikliklerde kırılmamalı,
- sürekli yeniden tasarım gerektirmemeli,
- sistem bağımlılıklarını bozmamalıdır.

---

# Flexibility (Esneklik) Nedir?

Flexibility:

> Yeni ihtiyaçların modele minimum maliyetle eklenebilmesidir.

Esnek modeller:

- yeni veri kaynaklarını,
- yeni iş kurallarını,
- yeni kullanım senaryolarını

kolay adapte edebilir.

---

# Kararlılık ve Esneklik Arasındaki Denge

Burada önemli bir trade-off vardır.

| Yaklaşım | Avantaj | Dezavantaj |
|---|---|---|
| Aşırı katı model | Tutarlılık sağlar | Değişiklik maliyeti yüksektir |
| Aşırı esnek model | Adaptasyon kolaydır | Karmaşıklık artabilir |

Amaç:

```text
Hem dayanıklı
hem de uyarlanabilir
bir model kurmaktır.
```

---

# Fatura Sistemi Örneği

Önceki invoice örneğini tekrar düşünelim.

Başlangıçta sistemde şu iş kuralı olsun:

```text
Her invoice_id benzersiz olmalıdır.
```

Bu oldukça mantıklı görünür.

Örneğin:

| invoice_id | amount |
|---|---|
| 1001 | 500 |
| 1002 | 900 |

---

# Problem Ne Zaman Ortaya Çıkıyor?

Şirket başka bir ülkede yeni bir invoicing sistemi kullanmaya başlıyor.

Yeni sistemde de:

```text
invoice_id = 1001
```

oluşabiliyor.

Artık elimizde:

| source_system | invoice_id |
|---|---|
| EU_SYSTEM | 1001 |
| US_SYSTEM | 1001 |

şeklinde iki farklı kayıt var.

---

# Katı Tasarımın Problemi

Eğer model başta yalnızca:

```text
invoice_id UNIQUE
```

şeklinde tasarlandıysa:

- primary key yapıları değişebilir,
- ETL süreçleri kırılabilir,
- downstream sistemler etkilenebilir,
- migration maliyeti oluşabilir.

Yani küçük görünen değişiklik:

```text
Büyük sistemsel maliyetlere dönüşebilir.
```

---

# Daha Esnek Tasarım

Eğer başlangıçta modele:

```text
source_system
```

alanı eklenmiş olsaydı yapı şöyle olurdu:

| source_system | invoice_id |
|---|---|
| EU_SYSTEM | 1001 |
| US_SYSTEM | 1001 |

Bu durumda unique yapı:

```text
(source_system, invoice_id)
```

kombinasyonu üzerinden kurulabilir.

---

# Esnek Tasarımın Avantajı

Bu yaklaşım:

✓ yeni sistemleri kolay ekler  
✓ mevcut modeli bozmaz  
✓ downstream sistemleri korur  
✓ migration maliyetini azaltır  
✓ ölçeklenebilirliği artırır  

---

# Veri Modellemede Önemli Gerçek

```text
Bugünkü ihtiyaçlar geleceğin garantisi değildir.
```

Bu nedenle veri modeli:

- yalnızca bugünü değil,
- olası gelecek senaryolarını da

olabildiğince düşünmelidir.

---

# Stability vs Flexibility Görselleştirmesi

```text
                İdeal Veri Modeli
                         │
        ┌────────────────┴────────────────┐
        │                                 │
        ↓                                 ↓

   Stability                        Flexibility
   (Kararlılık)                     (Esneklik)

- Dayanıklılık                  - Adaptasyon Yeteneği
- Tutarlılık                    - Genişleyebilirlik
- Güvenilirlik                  - Değişime Açıklık
```

```text
Teknik ekip ile iş ekipleri arasında ortak dil oluşturur.
```

Bu sayede:

- yanlış anlamalar azalır,
- veri güveni artar,
- iş birliği kolaylaşır.

---

# 1. Conceptual Data Model  
# Kavramsal Veri Modeli

İlk aşama:

```text
Conceptual Diagram
```

yani kavramsal diyagramdır.

Bu aşamada model:

- yüksek seviyelidir,
- teknoloji bağımsızdır,
- iş süreçlerine odaklanır,
- temel ilişkileri gösterir.

---

# Kavramsal Modelin Amacı

Conceptual model şu sorulara cevap verir:

| Soru | Amaç |
|---|---|
| Sistemde hangi varlıklar var? | Entity’leri belirlemek |
| Bu varlıklar nasıl ilişkili? | Relationship’leri anlamak |
| İş süreçleri nasıl çalışıyor? | Business flow oluşturmak |
| Temel veri akışı nedir? | Genel yapıyı görmek |

---

# Teknoloji Bağımsız Tasarım

Bu aşamada henüz:

- PostgreSQL,
- Snowflake,
- BigQuery,
- MySQL,
- MongoDB

gibi platform detayları düşünülmez.

Amaç:

```text
İş dünyasını veri perspektifinden modellemektir.
```

---

# Business Stakeholder Geri Bildirimi

Kavramsal model oluşturulduktan sonra:

- business specialist’ler,
- data owner’lar,
- son kullanıcılar,
- domain expert’ler

ile birlikte değerlendirme yapılır.

Çünkü amaç:

```text
Modelin iş dünyasını doğru temsil ettiğinden emin olmaktır.
```

---

# Basit Sipariş Yönetim Sistemi

Aşağıdaki yapı basit bir sipariş yönetim sisteminin kavramsal modelidir.

---

# Conceptual Diagram (Kavramsal Diyagram)

```text
Customer
    │
    │ places
    ▼
Order
    │
    ├──────────────► Order Item
    │                     │
    │                     │ represents
    │                     ▼
    │                  Product
    │
    │ generates
    ▼
Invoice
    │
    ▼
Invoice Item
```

---

# Diyagramın Yorumlanması

Bu modelde:

| İlişki | Açıklama |
|---|---|
| Customer → Order | Müşteri sipariş verir |
| Order → Order Item | Sipariş ürün kalemlerinden oluşur |
| Order Item → Product | Her kalem bir ürünü temsil eder |
| Order → Invoice | Siparişten fatura oluşur |
| Invoice → Invoice Item | Fatura kalemleri oluşur |

---

# Kavramsal Modelin Gücü

Bu aşamada henüz:

- veri tipi,
- primary key,
- foreign key,
- indexing,
- storage

gibi teknik detaylar yoktur.

Buna rağmen:

```text
Sistemin iş mantığı açık şekilde anlaşılabilir.
```

İşte conceptual modelin gücü budur.

---

# Kavramsal Modelin Özellikleri

| Özellik | Açıklama |
|---|---|
| High-level | Yüksek seviyeli |
| Business-oriented | İş odaklı |
| Technology-agnostic | Teknolojiden bağımsız |
| Relationship-focused | İlişki odaklı |
| Collaborative | İş birliği gerektirir |

# Mantıksal Modelde Eklenen Detaylar

Bu aşamada artık şunlar görünür hale gelir:

---

# Logical Diagram Örneği

```text
Customer
─────────────────────────
customer_id      INT
customer_name    VARCHAR


Order
─────────────────────────
order_id         INT
invoice_id       VARCHAR
customer_id      INT


Order Item
─────────────────────────
order_id         INT
product_id       INT


Product
─────────────────────────
product_id       INT
product_name     VARCHAR


Invoice
─────────────────────────
invoice_id       VARCHAR
invoice_date     DATE


Invoice Item
─────────────────────────
invoice_id       VARCHAR
product_id       INT
```

---

# Conceptual vs Logical Model

| Conceptual Model | Logical Model |
|---|---|
| İş odaklıdır | Teknik detay içerir |
| Entity gösterir | Table gösterir |
| Relationship odaklıdır | Column odaklıdır |
| Teknolojiden bağımsızdır | DBMS’e yaklaşır |
| Basit yapıdadır | Daha detaylıdır |

---

# Three-Schema Architecture  
# Üç Şema Mimarisi

1970’lerin ortasında:

```text
ANSI/SPARC
```

tarafından standartlaştırılan önemli bir yaklaşım ortaya çıktı:

# Three-Schema Architecture

---

# Üç Katmanlı Mimari

Bu mimari üç seviyeden oluşur:

```text
External Schema
        ↑
Conceptual Schema
        ↑
Internal Schema
```

---

# 1. Internal Schema  
# İç Şema

Internal schema:

```text
Verinin fiziksel olarak nasıl saklandığını tanımlar.
```

Örneğin:

- partitioning,
- clustering,
- storage format,
- indexing,
- table type

gibi konular burada yer alır.

---

# 2. Conceptual Schema  
# Kavramsal Şema

Conceptual schema:

```text
Paylaşılan mantıksal veri görünümünü temsil eder.
```

Burada:

- tablolar,
- kolonlar,
- ilişkiler,
- iş kuralları

tanımlanır.

Bu katman:

```text
İş dünyası ile teknik dünya arasındaki köprüdür.
```

---

# 3. External Schema  
# Dış Şema

External schema:

```text
Kullanıcılara özel veri görünümüdür.
```

Örneğin:

- BI view’ları,
- data mart’lar,
- reporting layer,
- kullanıcı bazlı görünümler

bu katmanda bulunur.

---

# Three-Schema Architecture Görselleştirmesi

```text
Business Users / Applications
                │
                ▼
        External Schema
                │
                ▼
       Conceptual Schema
                │
                ▼
         Internal Schema
                │
                ▼
          Physical Storage
```

---

# Bu Mimari Neden Önemlidir?

Bu ayrım sayesinde:

✓ kullanıcılar fiziksel değişikliklerden etkilenmez  
✓ storage optimizasyonları transparan yapılabilir  
✓ veri güvenliği sağlanabilir  
✓ farklı kullanıcılar farklı görünüm alabilir  

---

# Internal Schema Gerçek Hayatta Nedir?

Modern sistemlerde internal schema çoğu zaman:

```text
DDL (Data Definition Language)
```

olarak görülür.

Örneğin:

```sql
CREATE TABLE
CREATE INDEX
PARTITION BY
CLUSTER BY
```

ifadeleri internal schema’nın parçasıdır.

---

# Conceptual Schema Neden Kritik?

Kitap özellikle şu noktayı vurguluyor:

```text
Conceptual schema kontrollü şekilde değiştirilmelidir.
```

Çünkü bu katman:

- stakeholder’lar ile onaylanmış,
- iş mantığını temsil eden,
- merkezi veri yapısıdır.

---

# Kontrolsüz Değişiklik Problemi

Eğer herkes:

- tablo eklerse,
- kolon silerse,
- isim değiştirirse

zamanla:

# ERD Nedir?

ERD (Entity Relationship Diagram):

> Veri yapılarının ve bu yapılar arasındaki ilişkilerin görsel gösterimidir.

Temel amacı:

```text
Veri modelini anlaşılır hale getirmek
```

ve sistemin iş mantığını görselleştirmektir.

---

# ERD Neden Önemlidir?

İyi bir ERD:

✓ veri yapısını netleştirir  
✓ business logic’i görünür hale getirir  
✓ veri tekrarını azaltır  
✓ ilişkileri anlamayı kolaylaştırır  
✓ database tasarımını yönlendirir  

---

# ERD’nin Temel Bileşenleri

Bir ERD genellikle üç temel bileşenden oluşur:

| Bileşen | Açıklama |
|---|---|
| Entity | Gerçek dünyadaki nesne |
| Attribute | Entity’nin özelliği |
| Relationship | Entity’ler arasındaki ilişki |

---

# 1. Entity Nedir?

Entity:

> Gerçek dünyada bağımsız olarak var olan nesne veya kavramdır.

Örnekler:

- Customer
- Product
- Order
- Employee
- Invoice

---

# Entity Örneği

```text
┌─────────────────────┐
│ Customer            │
├─────────────────────┤
│ customer_id         │
│ customer_name       │
│ email               │
└─────────────────────┘
```

Burada:

```text
Customer
```

bir entity’dir.

---

# 2. Attribute Nedir?

Attribute:

> Entity hakkında bilgi tutan alanlardır.

Örneğin:

| Attribute | Açıklama |
|---|---|
| customer_id | Müşteri kimliği |
| customer_name | Müşteri adı |
| email | E-posta adresi |

---

# 3. Relationship Nedir?

Relationship:

> Entity’ler arasındaki bağlantıdır.

Örneğin:

```text
Customer places Order
```

Burada:

- Customer
ve
- Order

arasında ilişki vardır.

---

# Cardinality Nedir?

Cardinality:

> İki entity arasındaki ilişkinin sayısal yapısını ifade eder.

Yani:

```text
Bir kayıt diğerinden kaç tane olabilir?
```

sorusuna cevap verir.

---

# Temel Relationship Türleri

ERD’de üç temel ilişki tipi vardır:

| Relationship Türü | Açıklama |
|---|---|
| One-to-One (1:1) | Bir kayıt yalnızca bir kayıtla ilişkili |
| One-to-Many (1:N) | Bir kayıt birçok kayıtla ilişkili |
| Many-to-Many (N:M) | Birçok kayıt birçok kayıtla ilişkili |

---

# 1. One-to-One (1:1)

One-to-One ilişki:

> Bir kaydın yalnızca bir başka kayıtla ilişkili olmasıdır.

---

# Gerçek Hayat Örneği

Bir kişinin yalnızca bir pasaportu olsun.

```text
Person  ───── Passport
```

---

# One-to-One Görselleştirme

```text
┌──────────────┐        ┌──────────────┐
│ Person       │        │ Passport     │
├──────────────┤        ├──────────────┤
│ person_id PK │───────▶│ passport_id  │
│ full_name    │        │ person_id FK │
└──────────────┘        └──────────────┘
```

---

# One-to-One Ne Zaman Kullanılır?

Genellikle:

- güvenlik bilgileri,
- kullanıcı ayarları,
- detay tabloları,
- opsiyonel bilgiler

için kullanılır.

---

# One-to-One Avantajları

✓ tabloyu sadeleştirir  
✓ güvenlik izolasyonu sağlar  
✓ opsiyonel verileri ayırır  
✓ performansı iyileştirebilir  

---

# One-to-One Dezavantajları

✗ gereksiz parçalanma oluşturabilir  
✗ fazla JOIN ihtiyacı doğurabilir  

---

# 2. One-to-Many (1:N)

En yaygın ilişki tipidir.

Anlamı:

> Bir kayıt birçok kayıtla ilişkili olabilir.

---

# Gerçek Hayat Örneği

Bir müşteri birçok sipariş verebilir.

```text
Customer ─────< Order
```

---

# One-to-Many Görselleştirme

```text
┌──────────────────┐
│ Customer         │
├──────────────────┤
│ customer_id PK   │
│ customer_name    │
└──────────────────┘
          │
          │ 1
          │
          ▼
          N
┌──────────────────┐
│ Order            │
├──────────────────┤
│ order_id PK      │
│ customer_id FK   │
│ order_date       │
└──────────────────┘
```

---

# Relationship’in Mantığı

Burada:

- bir customer
- birçok order verebilir.

Ancak:

```text
Her order yalnızca bir customer’a aittir.
```

---

# SQL Perspektifi

Bu ilişki çoğu zaman:

```sql
customer_id
```

foreign key’i ile kurulur.

---

# One-to-Many En Çok Nerede Görülür?

| Parent | Child |
|---|---|
| Customer | Orders |
| Category | Products |

| Country | Cities |

---

# One-to-Many Avantajları

✓ normalize yapı sağlar  
✓ veri tekrarını azaltır  
✓ ölçeklenebilir yapı oluşturur  
✓ business process’leri doğal temsil eder  

---

# 3. Many-to-Many (N:M)

Many-to-Many ilişki:

> Birçok kaydın birçok kayıtla ilişkili olmasıdır.

---

# Gerçek Hayat Örneği

Bir öğrenci birçok ders alabilir.

Bir dersin de birçok öğrencisi olabilir.

```text
Student >────< Course
```

---

# Neden Direkt Kurulamaz?

Relational database’lerde many-to-many ilişki doğrudan kurulmaz.

Araya:

```text
Junction Table
```

eklenir.

---

# Many-to-Many Yapısı

```text
┌──────────────────┐
│ Student          │
├──────────────────┤
│ student_id PK    │
│ student_name     │
└──────────────────┘


┌──────────────────┐
│ Enrollment       │
├──────────────────┤
│ student_id FK    │
│ course_id FK     │
└──────────────────┘


┌──────────────────┐
│ Course           │
├──────────────────┤
│ course_id PK     │
│ course_name      │
└──────────────────┘
```

---

# Junction Table Nedir?

Junction Table:

> İki tablo arasındaki many-to-many ilişkiyi çözmek için kullanılan ara tablodur.

Başka isimleri:

- bridge table
- associative entity
- mapping table

# Many-to-Many Nerelerde Kullanılır?

# Primary Key ve Foreign Key  
# Veri İlişkilerinin Temeli

Relational database sistemlerinin en önemli yapı taşlarından biri:

```text
Primary Key (PK)
ve
Foreign Key (FK)
```

kavramlarıdır.

Bu yapılar:

- veri tutarlılığını sağlar,
- tablolar arası ilişki kurar,
- duplicate veriyi önler,
- relational model’in temelini oluşturur.

---

# Neden Key Yapıları Gereklidir?

Bir veritabanında milyonlarca kayıt olabilir.

Örneğin:

| customer_name |
|---|
| Ahmet |
| Ahmet |
| Mehmet |

Burada problem şudur:

```text
Hangi Ahmet?
```

İşte key yapıları bu problemi çözer.

---

# Primary Key (PK) Nedir?

Primary Key:

> Bir tablodaki her satırı benzersiz şekilde tanımlayan alandır.

---

# Primary Key’in Temel Özellikleri

Bir primary key:

✓ benzersiz olmalıdır  
✓ NULL olamaz  
✓ değişmemesi tercih edilir  
✓ her satırda bulunmalıdır  

---

# Primary Key Örneği

```text
┌──────────────────────────┐
│ Customer                 │
├──────────────────────────┤
│ customer_id PK           │
│ customer_name            │
│ email                    │
└──────────────────────────┘
```

---

# Neden customer_name PK Yapılmaz?

Çünkü:

- aynı isimde müşteriler olabilir,
- isim değişebilir,
- yazım farklılıkları oluşabilir.

Örneğin:

| customer_id | customer_name |
|---|---|
| 101 | Ahmet Yılmaz |
| 102 | Ahmet Yılmaz |

Burada:

```text
customer_name benzersiz değildir.
```

---

# İdeal Primary Key Nasıl Olmalı?

İdeal PK:

| Özellik | Açıklama |
|---|---|
| Stable | Değişmemeli |
| Unique | Benzersiz olmalı |
| Short | Küçük boyutlu olmalı |
| Simple | Basit olmalı |

---

# Surrogate Key vs Natural Key

Primary key’ler genellikle ikiye ayrılır:

| Tür | Açıklama |
|---|---|
| Natural Key | Gerçek iş verisinden oluşur |
| Surrogate Key | Yapay olarak üretilir |

---

# Natural Key Örneği

Örneğin:

```text
TC Kimlik No
```

bir natural key olabilir.

Çünkü:

- gerçek dünyada vardır,
- doğal olarak benzersizdir.

---

# Natural Key Problemleri

Ancak natural key’ler bazen problem oluşturabilir:

| Problem | Açıklama |
|---|---|
| Değişebilir | Email değişebilir |
| Uzun olabilir | String key maliyetlidir |
| Gizlilik problemi | Sensitive data olabilir |
| Bölgesel farklılık | Her ülkede aynı yapı yok |

---

# Surrogate Key Nedir?

Surrogate key:

> Sistem tarafından yapay olarak üretilen benzersiz anahtardır.

Genellikle:

```text
INTEGER AUTO_INCREMENT
UUID
SEQUENCE
```

şeklinde oluşturulur.

---

# Surrogate Key Örneği

| customer_id | customer_name |
|---|---|
| 1001 | Ahmet |
| 1002 | Mehmet |

Burada:

```text
customer_id
```

surrogate key’dir.

---

# Modern Data Warehouse Yaklaşımı

Modern veri ambarlarında çoğu zaman:

```text
Surrogate Key tercih edilir.
```

Çünkü:

✓ daha stabil  
✓ daha performanslı  
✓ daha yönetilebilir  
✓ daha güvenlidir  

---

# Composite Primary Key

Bazı durumlarda birden fazla kolon birlikte PK olabilir.

Buna:

```text
Composite Key
```

denir.

---

# Composite Key Örneği

```text
Enrollment Table
```

| student_id | course_id |
|---|---|
| 1 | 101 |
| 1 | 102 |

Burada:

```text
(student_id, course_id)
```

birlikte unique’tir.

---

# SQL ile Primary Key

```sql
CREATE TABLE customers (
    customer_id INT PRIMARY KEY,
    customer_name VARCHAR(255)
);
```

---

# Foreign Key (FK) Nedir?

Foreign Key:

> Bir tablodaki alanın başka bir tablonun primary key’ine referans vermesidir.

---

# Foreign Key’in Amacı

Foreign key:

✓ tablolar arası ilişki kurar  
✓ referential integrity sağlar  
✓ orphan record oluşmasını engeller  

---

# Foreign Key Örneği

```text
┌──────────────────────┐
│ Customer             │
├──────────────────────┤
│ customer_id PK       │
│ customer_name        │
└──────────────────────┘
            │
            │
            ▼
┌──────────────────────┐
│ Orders               │
├──────────────────────┤
│ order_id PK          │
│ customer_id FK       │
│ order_date           │
└──────────────────────┘
```

---

# Bu İlişkinin Anlamı

Burada:

```text
Orders.customer_id
```

alanı:

```text
Customer.customer_id
```

alanına bağlıdır.

Yani:

```text
Her order bir customer’a ait olmalıdır.
```

---

# Referential Integrity

Foreign key’lerin temel amacı:

```text
Referential Integrity
```

sağlamaktır.

---

# Referential Integrity Nedir?

Referential integrity:

> İlişkili verilerin tutarlı kalmasını sağlayan kuraldır.

Örneğin:

```text
Var olmayan customer için order oluşturulamaz.
```

---

# Hatalı Veri Örneği

Customers tablosu:

| customer_id |
|---|
| 1 |
| 2 |

Orders tablosu:

| order_id | customer_id |
|---|---|
| 100 | 999 |

Burada:

```text
customer_id = 999
```

customers tablosunda olmadığı için:

```text
referential integrity bozulur.
```

---

# SQL ile Foreign Key

```sql
CREATE TABLE orders (
    order_id INT PRIMARY KEY,
    customer_id INT,
    
    FOREIGN KEY (customer_id)
    REFERENCES customers(customer_id)
);
```

---

# Foreign Key’in Avantajları

| Avantaj | Açıklama |
|---|---|
| Veri tutarlılığı | Hatalı ilişkiyi engeller |
| Güvenilirlik | Sistem doğruluğunu artırır |
| Relationship yönetimi | Tablolar bağlanır |
| Veri kalitesi | Orphan kayıtları engeller |

---

# Orphan Record Nedir?

Orphan record:

> Referans verdiği parent kayıt olmayan child kayıttır.

Örneğin:

```text
Order var
ama customer yok.
```

Bu durum veri kalitesini bozar.

---

# Normalization  
# Normalizasyon

Normalization (Normalizasyon):

> Veriyi organize ederek tekrarları azaltma ve veri tutarlılığını artırma sürecidir.

Relational database tasarımının en önemli kavramlarından biridir.

Temel amacı:

✓ veri tekrarını azaltmak  
✓ veri tutarsızlığını önlemek  
✓ bakım maliyetini düşürmek  
✓ veri bütünlüğünü artırmak  

---

# Neden Normalization Gereklidir?

Kötü tasarlanmış veri yapıları:

- duplicate data,
- update anomaly,
- insert anomaly,
- delete anomaly

gibi ciddi problemlere yol açar.

---

# Veri Tekrarı Problemi

Örneğin aşağıdaki tabloyu düşünelim:

| order_id | customer_name | customer_city | product_name |
|---|---|---|---|
| 1001 | Ahmet | İstanbul | Laptop |
| 1002 | Ahmet | İstanbul | Mouse |
| 1003 | Ahmet | İstanbul | Keyboard |

Burada:

```text
Ahmet
İstanbul
```

bilgileri sürekli tekrar ediyor.

---

# Bu Yapının Problemleri

| Problem | Açıklama |
|---|---|
| Fazladan storage | Aynı veri tekrar ediyor |
| Update problemi | Şehir değişirse her satır güncellenmeli |
| Tutarsızlık | Bazı satırlar unutulabilir |
| Bakım zorluğu | Yönetim karmaşıklaşır |

---

# Anomaly Kavramı

Normalization’ın temel amacı:

```text
Data anomaly’leri azaltmaktır.
```

---

# 1. Update Anomaly

Bir veri birçok yerde tekrar ediyorsa:

```text
Hepsinin birlikte güncellenmesi gerekir.
```

Örneğin:

| order_id | customer_city |
|---|---|
| 1001 | İstanbul |
| 1002 | İstanbul |

Müşteri Ankara’ya taşınırsa:

```text
Tüm kayıtlar güncellenmelidir.
```

Bir satır unutulursa:

```text
Veri tutarsızlığı oluşur.
```

---

# 2. Insert Anomaly

Bazı yapılarda:

```text
Eksik bilgiyle veri eklenemez.
```

Örneğin:

Yeni customer var ama henüz order vermemiş.

Ancak yapı order odaklıysa:

```text
Customer kaydı eklenemeyebilir.
```

# Normalization’ın Temel Mantığı

# Normal Forms (NF)

Normalization seviyeleri:

| Kısaltma | Açılım |
|---|---|
| 1NF | First Normal Form |
| 2NF | Second Normal Form |
| 3NF | Third Normal Form |
| BCNF | Boyce-Codd Normal Form |
| 4NF | Fourth Normal Form |
| 5NF | Fifth Normal Form |

Pratikte en yaygın kullanılanlar:

```text
1NF, 2NF, 3NF
```

olur.

---

# 1NF — First Normal Form  
# Birinci Normal Form

Bir tablo:

```text
Atomic (bölünemez) değerler içeriyorsa
```

1NF’dedir.

---

# Atomic Value Nedir?

Atomic:

> Daha küçük mantıklı parçalara bölünmeyen veri.

---

# 1NF İhlali Örneği

| customer_id | phone_numbers |
|---|---|
| 1 | 555-1111, 555-2222 |

Burada:

```text
Bir hücrede birden fazla değer var.
```

Bu:

```text
1NF ihlalidir.
```

---

# 1NF Doğru Yapı

| customer_id | phone_number |
|---|---|
| 1 | 555-1111 |
| 1 | 555-2222 |

Artık:

```text
Her hücre tek değer içeriyor.
```

---

# 1NF Temel Kuralları

Bir tablo:

✓ atomic değer içermeli  
✓ tekrar eden grup içermemeli  
✓ her satır benzersiz olmalı  

---

# 2NF — Second Normal Form  
# İkinci Normal Form

2NF için:

```text
Önce 1NF sağlanmalıdır.
```

Ardından:

```text
Partial dependency olmamalıdır.
```

---

# Partial Dependency Nedir?

Bir kolonun:

```text
Composite key’in yalnızca bir kısmına bağlı olmasıdır.
```

---

# 2NF İhlali Örneği

| student_id | course_id | student_name |
|---|---|---|
| 1 | 101 | Ahmet |
| 1 | 102 | Ahmet |

Primary key:

```text
(student_id, course_id)
```

Ancak:

```text
student_name
```

yalnızca:

```text
student_id
```

alanına bağlı.

Bu:

```text
Partial dependency
```

oluşturur.

---

# 2NF Çözümü

Tabloyu ayırırız.

---

# Students Tablosu

| student_id | student_name |
|---|---|
| 1 | Ahmet |

---

# Enrollment Tablosu

| student_id | course_id |
|---|---|
| 1 | 101 |
| 1 | 102 |

---

# 2NF’nin Mantığı

```text
Her non-key kolon
primary key’in tamamına bağlı olmalıdır.
```

---

# 3NF — Third Normal Form  
# Üçüncü Normal Form

3NF için:

```text
Önce 2NF sağlanmalıdır.
```

Sonra:

```text
Transitive dependency olmamalıdır.
```

---

# Transitive Dependency Nedir?

Bir kolonun:

```text
Başka bir non-key kolona bağlı olmasıdır.
```

---

# 3NF İhlali Örneği

| employee_id | department_id | department_name |
|---|---|---|
| 1 | 10 | Finance |
| 2 | 20 | HR |

Burada:

```text
department_name
```

aslında:

```text
department_id
```

alanına bağlıdır.

Yani:

```text
employee_id
    ↓
department_id
    ↓
department_name
```

şeklinde dolaylı bağımlılık vardır.

---

# 3NF Çözümü

---

# Employees Tablosu

| employee_id | department_id |
|---|---|
| 1 | 10 |
| 2 | 20 |

---

# Departments Tablosu

| department_id | department_name |
|---|---|
| 10 | Finance |
| 20 | HR |

---

# 3NF’nin Temel Mantığı

```text
Non-key kolonlar
sadece primary key’e bağlı olmalıdır.
```

# Normalization  
# Normalizasyon Süreci

Normalization (normalizasyon):

> Veriyi mantıksal olarak organize ederek tekrarları azaltmayı, veri tutarlılığını artırmayı ve veri anomalilerini önlemeyi amaçlayan sistematik veri modelleme sürecidir.

Relational database tasarımının en önemli teorik temellerinden biridir.

---

# Normalization’ın Temel Amacı

Bir veri sistemi büyüdükçe:

- veri tekrarları artabilir,
- veri tutarsızlıkları oluşabilir,
- bakım maliyeti yükselir,
- relationship’ler karmaşıklaşabilir.

Normalization’ın amacı bu problemleri kontrol altına almaktır.

---

# Basit Tanım

Normalization aslında şunu yapar:

```text
Veriyi mantıklı parçalara ayırır.
```

Bunu yaparken:

✓ redundancy azaltılır  
✓ veri tutarlılığı korunur  
✓ veri ilişkileri netleşir  
✓ update problemleri önlenir  

---

# Neden Normalization Gereklidir?

Düşük kaliteli veri modellerinde genellikle şu problemler görülür:

| Problem | Açıklama |
|---|---|
| Duplicate Data | Aynı veri tekrar tekrar saklanır |
| Update Anomaly | Veri güncellemesi tutarsız hale gelir |
| Insert Anomaly | Veri eklemek zorlaşır |
| Delete Anomaly | Veri silinince başka bilgi kaybolabilir |

---

# Gerçek Hayat Analojisi

Bir şirket düşünelim.

Her çalışan bilgisi:

- maaş tablosunda,
- proje tablosunda,
- izin tablosunda,
- performans tablosunda

ayrı ayrı tekrar ediyorsa:

```text
Veri kaosu oluşur.
```

Çalışanın soyadı değiştiğinde:

```text
Tüm tablolar güncellenmek zorunda kalır.
```

Bu durum:

```text
Tutarsız veri üretir.
```

Normalization bu problemi çözmeye çalışır.

---

# Normalization Sürecinin Temel Mantığı

Kitapta belirtildiği gibi normalization:

```text
Tek bir kaynaktan başlar.
```

Ardından modelleyici:

- veri bağımlılıklarını analiz eder,
- hangi bilgilerin birlikte tutulması gerektiğini inceler,
- tekrar eden verileri ayırır,
- ilişkileri netleştirir.

---

# Normalization Sürecinin Görsel Mantığı

Kitaptaki şeklin mantığı şöyledir:

```text
Unnormalized Data
        ↓
Remove Repeating Groups
        ↓
1NF
        ↓
Remove Partial Dependencies
        ↓
2NF
        ↓
Remove Transitive Dependencies
        ↓
3NF
```

---

# Normalization Süreci Diyagramı

```text
Unnormalized Form
        │
        ▼
Remove Repeating Groups
        │
        ▼
First Normal Form (1NF)
        │
        ▼
Remove Partial Dependencies
        │
        ▼
Second Normal Form (2NF)
        │
        ▼
Remove Transitive Dependencies
        │
        ▼
Third Normal Form (3NF)
```

# First Normal Form (1NF) ve Repeating Groups  
# Birinci Normal Form ve Tekrarlayan Gruplar

Normalization sürecinin ilk büyük adımı:

```text
Repeating groups’ları (tekrarlayan grupları) tespit edip ayırmaktır.
```

Bu adım:

```text
First Normal Form (1NF)
```

olarak bilinir.

---

# First Normal Form (1NF) Nedir?

Bir tablo:

✓ her hücrede tek bir değer içeriyorsa  
✓ tekrar eden kolon grupları barındırmıyorsa  
✓ veriler düzenli tablo yapısında tutuluyorsa  

1NF’e uygundur.

# Second Normal Form (2NF)  
# İkinci Normal Form

First Normal Form (1NF) ile:

```text
Tekrarlayan grupları kaldırmıştık.
```

Ancak veri modelimiz hâlâ tamamen optimize edilmiş değildir.

Çünkü:

```text
Tekrarlayan kolon grupları kaldırılmış olsa bile
tekrarlayan veri problemleri devam edebilir.
```

İşte Second Normal Form (2NF):

```text
Bu tür redundancy problemlerini azaltmayı hedefler.
```

---

# 2NF’in Temel Mantığı
